# PepsiCo (PEP) Stock Analysis & EDA Dashboard
## 📈 High-Performance Financial Analysis (2000-2026)

**Author:** Data Science Team
**Objective:** To conduct a professional exploratory data analysis (EDA) on PepsiCo's historical stock performance, identifying trends, patterns, and actionable financial insights.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import warnings

# Configuration
warnings.filterwarnings('ignore')
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams['figure.figsize'] = (12, 6)
```

```
# Loading the dataset
df = pd.read_csv('PEP.csv')
df['Date'] = pd.to_datetime(df['Date'])
df.sort_values('Date', inplace=True)
df.reset_index(drop=True, inplace=True)

print("Dataset Loaded Successfully!")
df.head()
```

## 🔍 1. Understanding the Data
We analyze the composition, distribution, and basic characteristics of our financial variables.


```
def summarize_data(df):
    """
    Provides a comprehensive summary of the dataset's structure, 
    missing values, and basic statistics.
    """
    print("--- INFO ---")
    print(df.info())
    print("\n--- NULL VALUES ---")
    print(df.isnull().sum())
    print("\n--- DUPLICATES ---")
    print(f"Total Duplicate Rows: {df.duplicated().sum()}")
    print("\n--- STATISTICAL SUMMARY ---")
    return df.describe()

summarize_data(df)
```

## 🛠️ 2. Data Preprocessing
Ensuring data integrity for technical analysis.


```
# Adding time-based features
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['Daily_Return'] = df['Close'].pct_change() * 100

print("New Features Engineered: Year, Month, Day, Daily_Return")
df.head()
```

## 📊 3. Distribution & Composition Analysis
Understanding how stock prices and volume are distributed over the timeframe.


```
plt.figure(figsize=(15, 5))

# Subplot 1: Distribution of Closing Prices
plt.subplot(1, 2, 1)
sns.histplot(df['Close'], kde=True, color='blue')
plt.title('Distribution of Closing Prices')

# Subplot 2: Distribution of Daily Returns
plt.subplot(1, 2, 2)
sns.histplot(df['Daily_Return'].dropna(), kde=True, color='green')
plt.title('Distribution of Daily Returns (%)')

plt.tight_layout()
plt.show()
```

## 🔗 4. Relationships & Comparisons
How Volume relates to Price and how prices compare across years.


```
# Relationship: Volume vs Price
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Volume', y='Close', hue='Year', palette='viridis', alpha=0.5)
plt.title('Closing Price vs Volume (Colored by Year)')
plt.show()

# Yearly Closing Price Average
plt.figure(figsize=(12, 6))
df.groupby('Year')['Close'].mean().plot(kind='bar', color='skyblue')
plt.title('Average Yearly Closing Price')
plt.ylabel('Price (USD)')
plt.show()
```

## 🎬 5. Animated EDA (Plotly)
Tracing the journey of PepsiCo stock across time.


```
# Animated Line Chart of Close Price over the Years
fig = px.line(df, x='Date', y='Close', title='PepsiCo Stock Price Trend (Animated)',
              labels={'Close': 'Price (USD)'}, 
              template='plotly_dark')

fig.update_xaxes(rangeslider_visible=True)
fig.show()

# Animated Area Chart of Volume
fig_vol = px.area(df, x='Date', y='Volume', title='PepsiCo Trading Volume Over Time',
                 color_discrete_sequence=['#ff7f0e'], template='plotly_dark')
fig_vol.show()
```

## 🔢 6. Statistical Analysis & Outliers
Identifying Moving Averages and anomalies.


```
# Moving Averages
df['SMA_50'] = df['Close'].rolling(window=50).mean()
df['SMA_200'] = df['Close'].rolling(window=200).mean()

# Visualization of Moving Averages
plt.figure(figsize=(15, 7))
plt.plot(df['Date'], df['Close'], label='Close', alpha=0.5)
plt.plot(df['Date'], df['SMA_50'], label='50-day SMA')
plt.plot(df['Date'], df['SMA_200'], label='200-day SMA', color='red')
plt.title('Close Price with Moving Averages (Trend Confirmation)')
plt.legend()
plt.show()

# Outlier Detection using Z-Score on Daily Returns
df['Return_ZScore'] = (df['Daily_Return'] - df['Daily_Return'].mean()) / df['Daily_Return'].std()
outliers = df[np.abs(df['Return_ZScore']) > 3]

print(f"Total Anomalous Returns Detected: {len(outliers)}")
outliers[['Date', 'Close', 'Daily_Return']].head()
```

## 🖥️ 7. Professional Interactive Dashboard
A unified view for investment professionals.


```
def create_dashboard(df):
    fig = make_subplots(rows=2, cols=2, 
                        subplot_titles=('Price Trend', 'Volume Analysis', 
                                        'Price Distribution', 'Volatility (Daily Return)'))

    # Subplot 1: Price
    fig.add_trace(go.Scatter(x=df['Date'], y=df['Close'], name='Close Price'), row=1, col=1)
    
    # Subplot 2: Volume
    fig.add_trace(go.Bar(x=df['Date'], y=df['Volume'], name='Volume'), row=1, col=2)
    
    # Subplot 3: Histogram
    fig.add_trace(go.Histogram(x=df['Close'], name='Price Hist'), row=2, col=1)
    
    # Subplot 4: Returns
    fig.add_trace(go.Scatter(x=df['Date'], y=df['Daily_Return'], name='Returns'), row=2, col=2)

    fig.update_layout(height=800, width=1100, title_text="PepsiCo Financial Dashboard", 
                      showlegend=False, template='plotly_dark')
    fig.show()

create_dashboard(df)
```
